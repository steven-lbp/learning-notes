### 1. 二分查找 / Binary Search(704)


```python

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        left = 0
        right = len(nums) - 1
        while left <= right:
            mid = (right - left)//2 + left
            if nums[mid] == target:
                return mid
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1
        return -1
```

### 2. 移除元素 / Remove Element(27)

给定整数数组 nums 和 val，原地移除所有等于 val 的元素，返回剩余元素数量。元素顺序可以改变，示例中的下划线表示无需关心的位置。

Given an integer array nums and an integer val, remove all occurrences of val in nums in-place. The order of the elements may be changed. Then return the number of elements in nums which are not equal to val.

- Example 1:

Input: nums = [3,2,2,3], val = 3

- Example 2:

```python

class Solution:
    def removeElement(self, nums: List[int], val: int) -> int:
        index = 0 # slow pointer
        for i in nums:
            if i != val:
                nums[index] = i
                index += 1
        return index
```

### 3. 有序数组的平方 / Squares of a Sorted Array(977)

示例一先把数组元素平方，得到 [16,1,0,9,100]，排序后为 [0,1,9,16,100]。示例二的平方排序结果为 [4,9,9,49,121]。

- Example 1:

Input: nums = [-4,-1,0,3,10]

Output: [0,1,9,16,100]

Explanation: After squaring, the array becomes [16,1,0,9,100].

After sorting, it becomes [0,1,9,16,100].

- Example 2:

Input: nums = [-7,-3,2,3,11]

Output: [4,9,9,49,121]


```python
class Solution:
    def sortedSquares(self, nums: List[int]) -> List[int]:
        left = 0
        right = len(nums) - 1
        n = len(nums)
        ans = [0] * n
        while True:
            if n == 0:
                break
            if abs(nums[left]) < abs(nums[right]):
                ans[n-1] = nums[right]**2
                right = right - 1
                
            else:
                ans[n-1] = nums[left]**2
                left = left + 1

            n = n-1
        return ans
```

### 5. 螺旋矩阵 II / Spiral Matrix II(59)


```python
class Solution:
    def generateMatrix(self, n: int) -> List[List[int]]:
        mat = [[0]*n for _ in range(n)]
        left, right = 0, n-1 
        top, bottom = 0, n-1
        val = 1
        while left<=right:
            # fill in every val in the top row
            for c in range(left, right+1):
                mat[top][c] = val
                val +=1
            top += 1

            # fill in every val in the right col
            for r in range(top, bottom+1):
                mat[r][right] = val
                val +=1
            right -= 1

            # fill in every val in the bottom row
            for c in range(right, left-1, -1):
                mat[bottom][c] = val
                val +=1
            bottom -= 1
            
            # fill in every val in the left row
            for r in range(bottom, top-1, -1):
                mat[r][left] = val
                val +=1
            left += 1
        return mat
```

### 2. 设计链表 / Design Linked List(707)


```python
class ListNode:
    def __init__(self, val = 0, next = None):
        self.val = val
        self.next = next

class MyLinkedList:

    def __init__(self, val = 0, next = None):
        self.head = None  # 会在添加节点后变成链表，不是空值
        self.size = 0

    def get(self, index: int) -> int:
        if index < 0 or index >= self.size:
            return -1
        curr = self.head
        for _ in range(index):
            curr = curr.next
        return curr.val

    def addAtHead(self, val: int) -> None:
        self.head = ListNode(val = val, next = self.head)
        self.size += 1

    def addAtTail(self, val: int) -> None:
        if not self.head:
            self.addAtHead(val)
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = ListNode(val = val, next = None)
        self.size += 1

    def addAtIndex(self, index: int, val: int) -> None:
        if index > self.size:
            return 
        if index <= 0:
            self.addAtHead(val)
            return
        curr = self.head
        for _ in range(index-1):
            curr = curr.next
        curr.next = ListNode(val = val, next = curr.next)
        self.size += 1

    def deleteAtIndex(self, index: int) -> None:
        if index >= self.size or index < 0:
            return 
        if index == 0:
            self.head = self.head.next
        else:
            curr = self.head
            for _ in range(index - 1):
                curr = curr.next
            curr.next = curr.next.next
        self.size -= 1
```

### 3. 反转链表 / Reverse Linked List(206)


```python
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        prev = None
        curr = head
        while curr:
            temp = curr.next
            curr.next = prev
            prev = curr
            curr = temp
        return prev
```

### 4. 两两交换链表节点 / Swap Nodes in Pairs(24)


```python
# draw a picture and it will be much easier
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode(next = head)
        prev, curr = dummy, head
        while curr and curr.next:
            # save ptrs
            nxtPair = curr.next.next
            second = curr.next

            # reverse this pair
            second.next = curr
            curr.next = nxtPair
            prev.next = second

            # update ptrs
            prev = curr
            curr = nxtPair
        return dummy.next
```

### 5. 删除链表倒数第 N 个节点 / Remove Nth Node From End of List(19)


```python
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        dummy = ListNode(next = head)
        left = dummy
        right = dummy
        for _ in range(n):
            right = right.next
        while right.next:
            left = left.next
            right = right.next
        left.next = left.next.next
        return dummy.next
```

### 6. 相交链表 / Intersection of Two Linked Lists(160)


```python
class Solution:
    def getIntersectionNode(self, headA: ListNode, headB: ListNode) -> Optional[ListNode]:
        lengthA, lengthB = 0, 0
        while headA:
            headA = headA.next
            lengthA +=1
        while headB:
            headB = headB.next
            lengthB +=1
        a, b = headA, headB
        if lengthA > lengthB:
            a, b = b, a
            lengthA, lengthB = lengthB, lengthA
        for _ in range(lengthB - lengthA):
            b = b.next
        while a:
            if a == b:  # not the value of a and b, but the linked list a and b
                return a
            else:
                a = a.next
                b = b.next
        return None
```

### 3. 逆波兰表达式求值 / Evaluate Reverse Polish Notation(150)


```python
class Solution:
    def evalRPN(self, tokens: List[str]) -> int:
        stack = []
        for str in tokens:
            if str in ["+", "-", "*", "/"]:
                b = int(stack.pop())
                a = int(stack.pop())
                if str == "+":
                    c = a+b
                elif str == "-":
                    c = a-b
                elif str == "*":
                    c = a*b
                else:
                    c = int(a/b)
                stack.append(c)
            else:
                stack.append(str)
        return int(stack[0])
```

### 4. 滑动窗口最大值 / Sliding Window Maximum(239)


```python
from collections import deque
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        que = deque()
        result = []
        for i in range(k): #先将前k的元素放进队列
            que.append(nums[i])
        result.append(max(list(que))) #result 记录前k的元素的最大值
        for i in range(k, len(nums)):
            que.popleft() #滑动窗口移除最前面元素
            que.append(nums[i])
            result.append(max(list(que))) #记录对应的最大值
        return result
```

### 2. 二叉树遍历：迭代 / Binary Tree Traversal(Iterate)


```python
# Preorder(mid, left, right) 144
class Solution:
    def preorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        res = []
        if not root:
            return res

        stack = [root]
        while stack:
            node = stack.pop()
            res.append(node.val)
            if node.right:  # right point first because it will be pop last
                stack.append(node.right)
            if node.left:  # # right point second because it will be pop first
                stack.append(node.left)
        return res

# Postorder(left, right, mid) 145
class Solution:
    def postorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        ans_rev = []  # the reverse array of answer(mid, right, left)
        if not root:
            return ans_rev

        stack = [root]
        while stack:
            node = stack.pop()
            ans_rev.append(node.val)
            if node.left:
                stack.append(node.left)
            if node.right:
                stack.append(node.right)
        return ans_rev[::-1]

# Inorder(left, mid, right) 94 (not a good solution)
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        stack = []
        res = []
        if not root:
            return res
        cur = root  # pointer
        while cur or stack:
            if cur:
                stack.append(cur)
                cur = cur.left
            else:
                cur = stack.pop()
                res.append(cur.val)
                cur = cur.right
        return res
```

### 3. 二叉树层序遍历 / Binary Tree Level Order Traversal(102)


```python
from collections import deque
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        if not root:
            return []
        queue = deque([root])
        ans = []
        
        while queue:
            level = []
            for _ in range(len(queue)):
                node = queue.popleft()
                level.append(node.val)
                if node.left:  # Add left child if it exists
                    queue.append(node.left)
                if node.right:  # Add right child if it exists
                    queue.append(node.right)
            ans.append(level)
        return ans
```

### 4. 翻转二叉树 / Invert Binary Tree(226)


```python
class Solution:
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        if not root: 
            return None

        queue = deque([root])
        
        while queue:
            node = queue.popleft()
            node.left, node.right = node.right, node.left
            if node.left:  # Add left child if it exists
                queue.append(node.left)
            if node.right:  # Add right child if it exists
                queue.append(node.right)
            
        return root
```

### 5. 对称二叉树 / Symmetric Tree


```python
from collections import deque
class Solution:
    def isSymmetric(self, root: Optional[TreeNode]) -> List[List[int]]:
        if not root:
            return []
        queue = deque([root.left, root.right])
        
        while queue:
            level = []
            for _ in range(len(queue)):
                node = queue.popleft()
                if node:
                    level.append(node.val)
                    queue.append(node.left)  # use None node to occupy position even there's no left tree
                    queue.append(node.right)
                else:
                    level.append(None)
            if level != level[::-1]:
                return False
            
        return True
```

### 6. 二叉树最大深度 / Maximum Depth of Binary Tree(104)


```python
class Solution:
    def getDepth(self, node):
        if not node:
            return 0
        leftheight = self.getDepth(node.left)
        rightheight = self.getDepth(node.right)
        height = 1 + max(leftheight, rightheight)
        return height
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        return self.getDepth(root)
```

### 7. 二叉树最小深度 / Minimum Depth of Binary Tree(111)


```python
from collections import deque
class Solution:    
    def minDepth(self, root: Optional[TreeNode]) -> int:
        if not root:
            return 0
        queue = deque([root])
        depth = 0
        while queue:
            depth += 1
            for _ in range(len(queue)):
                node = queue.popleft()
                if not node.left and not node.right:  # only when a node has no left and right subtree can it called the minimum depth
                    return depth
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
                
        return depth
```

### 1. 斐波那契数 / Fibonacci Number(509)


```python
class Solution:
    def fib(self, n: int) -> int:
        df = [0]*(n+1)
        df[0] = 0
        df[1] = 1
        for i in range(2, n+1):
            df[i] = df[i-1] + df[i-2]
        return df[n]
```

### 2. 爬楼梯 / Climbing Stairs(70)


```python
class Solution:
    def climbStairs(self, n: int) -> int:
        dp = [0]*(n+1)
        dp[0] = 0
        dp[1] = 1
        dp[2] = 2
        for i in range(3, n+1):
            dp[i] = dp[i-1] + dp[i-2]
        return dp[n]
```

到达第 i 级台阶有两种来源：

- 从第 i−1 级走一步。
- 从第 i−2 级走两步。

To reach step i, you can either:
- Come from step i-1 (1-step move).
- Come from step i-2 (2-step move).


### 3. 使用最小花费爬楼梯 / Min Cost Climbing Stairs(746)


```python
class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        n = len(cost)
        if n < 2:
            return cost[n]
        dp = [0]*(n+1)
        dp[1] = 0
        for i in range(2, n + 1):
            dp[i] = min(dp[i-2] + cost[i-2], dp[i-1]+cost[i-1])
        return dp[n]
```

### 4. 不同路径 / Unique Paths(62)


```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [[0]*n for _ in range(m)]
        for i in range(n):
            dp[0][i] = 1
        for i in range(m):
            dp[i][0] = 1

        for i in range(1, m):
            for j in range(1, n):
                dp[i][j] = dp[i-1][j] + dp[i][j-1]

        return dp[m-1][n-1]
```

### 5. 不同路径 II / Unique Paths2(63)


```python
class Solution:
    def uniquePathsWithObstacles(self, obstacleGrid: List[List[int]]) -> int:
        m = len(obstacleGrid)
        n = len(obstacleGrid[0])
        if obstacleGrid[m - 1][n - 1] == 1 or obstacleGrid[0][0] == 1:
            return 0
        dp = [[0]*n for _ in range(m)]

        for i in range(m):
            if obstacleGrid[i][0] == 0:
                dp[i][0] = 1
            else:
                break
        for j in range(n):
            if obstacleGrid[0][j] == 0:
                dp[0][j] = 1
            else:
                break

        for i in range(1, m):
            for j in range(1, n):
                if obstacleGrid[i][j] == 1:
                    continue
                dp[i][j] = dp[i-1][j] + dp[i][j-1]
        return dp[m-1][n-1]
```

### 6. 整数拆分 / Integer Break(343)


```python
class Solution:
    def integerBreak(self, n: int) -> int:
        dp = [0]* (n+1)
        dp[2] = 1
        for i in range(3, n+1):
            for j in range(1, i//2 + 1):
                dp[i] = max(dp[i], (i-j)*j, dp[i-j]*j)
        return dp[n]
```

### 7. 分割等和子集 / Partition Equal Subset Sum(416)


```python
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        
        total_sum = sum(nums)

        if total_sum % 2 != 0:
            return False

        target_sum = total_sum // 2
        dp = [[False] * (target_sum + 1) for _ in range(len(nums) + 1)]

        # 初始化第一列（空子集可以得到和为0）
        for i in range(len(nums) + 1):
            dp[i][0] = True

        for i in range(1, len(nums) + 1):
            for j in range(1, target_sum + 1):
                if j < nums[i - 1]:
                    # 当前数字大于目标和时，无法使用该数字
                    dp[i][j] = dp[i - 1][j]
                else:
                    # 当前数字小于等于目标和时，可以选择使用或不使用该数字
                    dp[i][j] = dp[i - 1][j] or dp[i - 1][j - nums[i - 1]]

        return dp[len(nums)][target_sum]
```

### 8. 最后一块石头的重量 II / Last Stone Weight II(1049)


```python
class Solution:
    def lastStoneWeightII(self, stones: List[int]) -> int:
        total_sum = sum(stones)
        target = total_sum // 2
        dp = [[False] * (target + 1) for _ in range(len(stones) + 1)]
        for i in range(len(stones) + 1):
            dp[i][0] = True

        for i in range(1, len(stones) + 1):
            for j in range(1, target + 1):
                if stones[i - 1] > j:
                    dp[i][j] = dp[i-1][j]
                else:
                    dp[i][j] = dp[i-1][j] or dp[i-1][j - stones[i - 1]]
        for i in range(target, -1, -1):
            if dp[len(stones)][i]:
                return total_sum - 2 * i
        
        return 0
```
